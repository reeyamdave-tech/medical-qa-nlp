### Importing Libraries

```python
import spacy
import scispacy
```

### Loading the Pre-Trained Model

```python
nlp = spacy.load("en_ner_bc5cdr_md")
```

## UPLOADING THE DATASET

```python
import pandas as pd

train_df = pd.read_csv('train.csv.zip')
print(train_df.head())
```

## MY TEXT DATA PROCESSING (METHOD 3)

```python
# Run NER on the Answer column instead
for answer in train_df['Answer'].head(10):
    doc = nlp(str(answer))

    # Identify and classify named entities
    entities = []
    for ent in doc.ents:
        entities.append((ent.text, ent.label_))

    # Print the named entities and their categories
    if entities:
        print("Entities found:")
        for entity, category in entities:
            print(f"{entity}: {category}")
        colors = {"DISEASE": "yellow", "CHEMICAL": "red"}
        colour_options = {"ents": ["DISEASE", "CHEMICAL"], "colors": colors}
        spacy.displacy.render(doc, style="ent", jupyter=True, options=colour_options)
    else:
        print("No entities found")
    print()
```

```python
# Run NER on the Question column instead
for question in train_df['Question'].head(10):
    doc = nlp(str(question))

    # Identify and classify named entities
    entities = []
    for ent in doc.ents:
        entities.append((ent.text, ent.label_))

    # Print the named entities and their categories
    if entities:
        print("Entities found:")
        for entity, category in entities:
            print(f"{entity}: {category}")
        colors = {"DISEASE": "yellow", "CHEMICAL": "red"}
        colour_options = {"ents": ["DISEASE", "CHEMICAL"], "colors": colors}
        spacy.displacy.render(doc, style="ent", jupyter=True, options=colour_options)
    else:
        print("No entities found")
    print()
```
